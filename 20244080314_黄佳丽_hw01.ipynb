```python
# 1.定义say_hi()函数
def say_hi():
    print("Hello, world!")

# 2.定义goat_pad()函数
def goat_pad(s):
    print("goat" + s + "goat")

# 3.定义print_n()函数
def print_n(s, n):
    for i in range(n):
        print(s)

say_hi()
goat_pad("hello")
print_n("hi", 3)
```

```python
# gcd
def gcd(m, n):
    m, n = abs(m), abs(n)
    while n != 0:
        m, n = n, m % n
    return m

# 评估数字对
print("gcd(2018, 2019) =", gcd(2018, 2019))
print("gcd(1200, 300) =", gcd(1200, 300))
print("gcd(5040, 60) =", gcd(5040, 60))

# 如果参数是负数
print("gcd(-2018, 2019) =", gcd(-2018, 2019))
print("gcd(2018, -2019) =", gcd(2018, -2019))
print("gcd(-2018, -2019) =", gcd(-2018, -2019))
```

```python
import math

# 1.euler_limit
def euler_limit(n):
    return (1 + 1 / n) ** n

print("euler_limit(1000) =", euler_limit(1000))
print("math.exp(1) =", math.exp(1))

# 2. euler_infinite_sum
def euler_infinite_sum(n):
    total = 0.0
    fact = 1.0
    for k in range(n):
        if k > 0:
            fact *= k
        total += 1 / fact
    return total

print("euler_infinite_sum(0) =", euler_infinite_sum(0))
print("euler_infinite_sum(1) =", euler_infinite_sum(1))
print("euler_infinite_sum(4) =", euler_infinite_sum(4))
print("math.exp(1) =", math.exp(1))

# 3.euler_approx
def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    fact = 1.0
    k = 0
    while abs(total - true_e) > epsilon:
        if k > 0:
            fact *= k
        total += 1 / fact
        k += 1
    return total

print("euler_approx(0.01) =", euler_approx(0.01))
print("euler_approx(0.0001) =", euler_approx(0.0001))

# 4.print_euler_sum_table
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

print_euler_sum_table(5)
```

```python
# 1.is_power_of_2
def is_power_of_2(n):
    if n <= 0:
        return False
    while n % 2 == 0:
        n = n // 2
    return n == 1

print("is_power_of_2(1) =", is_power_of_2(1))
print("is_power_of_2(2) =", is_power_of_2(2))
print("is_power_of_2(3) =", is_power_of_2(3))
print("is_power_of_2(4) =", is_power_of_2(4))
print("is_power_of_2(16) =", is_power_of_2(16))
print("is_power_of_2(18) =", is_power_of_2(18))

# 2.is_power
def is_power(b, n):
    if n <= 0:
        return False
    while n % b == 0:
        n = n // b
    return n == 1

print("is_power(2, 8) =", is_power(2, 8))
print("is_power(2, 9) =", is_power(2, 9))
print("is_power(3, 27) =", is_power(3, 27))
print("is_power(3, 28) =", is_power(3, 28))
print("is_power(5, 125) =", is_power(5, 125))
```
