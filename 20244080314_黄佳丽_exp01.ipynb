```python
# 1.score_level
def score_level(score):
    if score < 0 or score > 100:
        raise ValueError("分数必须在 0 到 100 之间")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

print(score_level(95))
print(score_level(85))
print(score_level(70))
print(score_level(50))

# 2.使用for和while，求1到100的和
total_for = 0
for i in range(1, 101):
    total_for += i
print("for 结果：", total_for)

total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1
print("while 结果：", total_while)

# 3.阶乘（递归版和迭代版）
def factorial_recursive(n):
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

for n in [0, 1, 5, 10]:
    print(n, factorial_recursive(n), factorial_iterative(n))

# 4.summary
def summary(values):
    return min(values), max(values), sum(values) / len(values)

values = [85, 58, 92, 76, 60, 89]
low, high, avg = summary(values)
print("最小值：", low)
print("最大值：", high)
print("平均值：", avg)

# 5.sum_all
def sum_all(*args):
    return sum(args)

print(sum_all())
print(sum_all(5))
print(sum_all(1, 2, 3, 4))
```

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

# 1.使用zip(),构建字典
name_score = dict(zip(names, scores))
for name, score in name_score.items():
    print(name, score)

# 2.使用enumerate()输出
for idx, (name, score) in enumerate(zip(names, scores), start=1):
    print(idx, name, score)

# 3.使用列表推导式
passed = [name for name, score in zip(names, scores) if score >= 60]
print("及格学生：", passed)

squares = [score ** 2 for score in scores]
print("成绩平方：", squares)

# 4.统计
count = len(scores)
avg = sum(scores) / count
max_score = max(scores)
min_score = min(scores)
pass_rate = len(passed) / count

print(f"人数：{count}")
print(f"平均分：{avg:.2f}")
print(f"最高分：{max_score}")
print(f"最低分：{min_score}")
print(f"及格率：{pass_rate:.2%}")

# 5.alias与copy
demo = [1, 2, 3]
alias = demo
copied = demo.copy()

alias[0] = 100
print("demo：", demo)
print("alias：", alias)
print("copied：", copied)

# 6.元组与zip长度不同
record = ("Alice", 85)
try:
    record[0] = "Bob"
except TypeError as e:
    print("元组不可修改：", e)

print(list(zip([1, 2, 3], ["a", "b"])))
```

```python
import os

# 1.查看工作目录并创建数据目录
print(os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)

file_path = os.path.join(data_dir, "students.txt")

# 2.写入数据
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

with open(file_path, "w", encoding="utf-8") as f:
    f.write(content)

# 读取并跳过表头
with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

for line in lines[1:]:
    print(line.strip())

# 3.编写parse_student(line)函数
def parse_student(line):
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError("字段数量不正确")
    name, score_str = parts
    if not name:
        raise ValueError("姓名不能为空")
    score = int(score_str)
    if score < 0 or score > 100:
        raise ValueError("成绩必须在 0 到 100 之间")
    return {"name": name, "score": score}

# 4.逐行处理并捕获异常
valid_records = []
invalid_count = 0

with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

for idx, line in enumerate(lines[1:], start=2):
    try:
        record = parse_student(line)
        valid_records.append(record)
    except ValueError as e:
        invalid_count += 1
        print(f"第 {idx} 行无效：{line.strip()}，原因：{e}")

print("有效记录数：", len(valid_records))
print("无效记录数：", invalid_count)

# 5.写入report.txt
valid_scores = [r["score"] for r in valid_records]
avg_score = sum(valid_scores) / len(valid_scores)
pass_count = sum(1 for s in valid_scores if s >= 60)
pass_rate = pass_count / len(valid_scores)

report_path = os.path.join(data_dir, "report.txt")
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数：{len(valid_records)}\n")
    f.write(f"无效记录数：{invalid_count}\n")
    f.write(f"平均分：{avg_score:.2f}\n")
    f.write(f"及格率：{pass_rate:.2%}\n")

with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())

# 6.pickle保存和读取
import pickle

pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)

with open(pkl_path, "rb") as f:
    loaded = pickle.load(f)

print(loaded == valid_records)

# 7.模块化
# 先创建student_utils.py，再导入
with open("student_utils.py", "w", encoding="utf-8") as f:
    f.write('''def parse_student(line):
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError("字段数量不正确")
    name, score_str = parts
    if not name:
        raise ValueError("姓名不能为空")
    score = int(score_str)
    if score < 0 or score > 100:
        raise ValueError("成绩必须在 0 到 100 之间")
    return {"name": name, "score": score}
''')

import importlib
import student_utils
importlib.reload(student_utils)

print(student_utils.parse_student("Alice,85"))
try:
    student_utils.parse_student("Grace,abc")
except ValueError as e:
    print("无效输入：", e)

# 8.FileNotFoundError
try:
    with open("not_exist.txt", "r", encoding="utf-8") as f:
        f.read()
except FileNotFoundError as e:
    print("文件不存在：", e)
```

```python
# 1.定义Student类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"

# 2.为有效记录创建对象
students = [Student(r["name"], r["score"]) for r in valid_records]
for s in students:
    print(s, "及格" if s.is_passed() else "不及格")

# 3.定义GraduateStudent类
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, topic={self.research_topic})"

gs = GraduateStudent("Alice", 85, "机器学习")
print(gs)
print(gs.is_passed())

# 4.对象复制
import copy

original = {"name": "Alice", "scores": [85, 90]}

direct = original
shallow = copy.copy(original)
deep = copy.deepcopy(original)

direct["scores"].append(100)
print("direct 修改后：", original)

original = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(original)
shallow["scores"].append(100)
print("shallow 修改后：", original)

original = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(original)
deep["scores"].append(100)
print("deep 修改后：", original)
```
