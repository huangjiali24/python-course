1.字符串的乐趣

```python
def is_palindrome(s): 
    clean_s = "".join(c.lower() for c in s if c != " ")
    return clean_s == clean_s[::-1]

def is_abecedarian(s):
    clean_s = "".join(c.lower() for c in s if c != " ")
    for i in range(len(clean_s) - 1):
        if clean_s[i] > clean_s[i+1]:
            return False
    return True

def remove_vowels(s):
    vowels = "aeiouAEIOU"
    return "".join(c for c in s if c not in vowels)

#进行测试
print("字符串的乐趣：")
print("is_palindrome('taco cat'):", is_palindrome("taco cat"))
print("is_abecedarian('abcd efgh xyz'):", is_abecedarian("abcd efgh xyz"))
print("remove_vowels('audio'):", remove_vowels("audio"))
```

2.列表的乐趣

```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是一个列表")
    return lst[::-1]

def is_sorted(seq):
    #使用zip将原序列和错位序列组合，单次循环进行比较
    for i in range(len(seq) - 1):
        if seq[i] > seq[i+1]:
            return False
    return True

#进行测试
print("列表的乐趣：")
print("list_reverse([1, 2, 3]):", list_reverse([1, 2, 3]))
print("is_sorted([1, 2, 2, 4]):", is_sorted([1, 2, 2, 4]))
print("is_sorted((3, 1, 2)):", is_sorted((3, 1, 2)))
```

3.作为向量的元组

```python
def vec_scalar_mult(t, s):
    #检查类型
    if not isinstance(t, tuple) or not isinstance(s, (int, float)):
        raise TypeError("第一个参数必须是元组，第二个参数必须是数字")
    if not all(isinstance(x, (int, float)) for x in t):
        raise TypeError("元组元素必须是数字")
    return tuple(x * s for x in t)

def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("输入必须均为元组")
    if len(t1) != len(t2):
        raise ValueError("两个元组长度必须一致")
    return sum(x * y for x, y in zip(t1, t2))

def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if not all(isinstance(row, tuple) for row in mx):
        return False
    if len(mx) == 0:
        return True 
    first_len = len(mx[0])
    for row in mx:
        if len(row) != first_len:
            return False
        if not all(isinstance(x, (int, float)) for x in row):
            return False
    return True

def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise TypeError("无效的矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    if len(mx) == 0 or len(mx[0]) != len(vec):
        raise ValueError("矩阵的列数必须等于向量的长度")
    
    #矩阵的每一行与向量做内积
    return tuple(vec_inner_product(row, vec) for row in mx)

#进行测试
print("元组向量与矩阵：")
print("vec_scalar_mult((1, 2), 3):", vec_scalar_mult((1, 2), 3))
print("vec_inner_product((1, 2), (3, 4)):", vec_inner_product((1, 2), (3, 4)))
mx = ((1, 2), (3, 4))
vec = (5, 6)
print("check_valid_mx:", check_valid_mx(mx))
print("mx_vec_mult:", mx_vec_mult(mx, vec))
```

4.更多向量的乐趣

```python
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or dim < 0:
            raise TypeError("维度必须是正整数")
        
        #如果未提供元素，默认为全0
        if elements is None:
            elements = [0] * dim
        else:
            if not isinstance(elements, (list, tuple)):
                raise TypeError("元素必须是列表或元组")
            if len(elements) != dim:
                raise ValueError("维度与元素个数不匹配")
            if not all(isinstance(x, (int, float)) for x in elements):
                raise TypeError("元素必须全是数字")
        
        self.dim = dim
        self.elements = tuple(elements)

    #用字符串表示
    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    #比较操作符
    def __eq__(self, other):
        if not isinstance(other, Vector): return NotImplemented
        self._check_dim_match(other)
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector): return NotImplemented
        self._check_dim_match(other)
        return self.elements < other.elements

    def __le__(self, other):
        if not isinstance(other, Vector): return NotImplemented
        self._check_dim_match(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        if not isinstance(other, Vector): return NotImplemented
        self._check_dim_match(other)
        return self.elements > other.elements

    def __ge__(self, other):
        if not isinstance(other, Vector): return NotImplemented
        self._check_dim_match(other)
        return self.elements >= other.elements

    #检查维度是否一致
    def _check_dim_match(self, other):
        if self.dim != other.dim:
            raise ValueError("两个向量的维度不同，无法比较或运算")

    #内积
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector 对象")
        self._check_dim_match(other)
        return sum(a * b for a, b in zip(self.elements, other.elements))

    #乘法运算符
    def __mul__(self, other):
        #向量*标量
        if isinstance(other, (int, float)):
            return Vector(self.dim, tuple(x * other for x in self.elements))
        #向量*向量
        elif isinstance(other, Vector):
            self._check_dim_match(other)
            return Vector(self.dim, tuple(a * b for a, b in zip(self.elements, other.elements)))
        else:
            raise TypeError("只能与数字或其他 Vector 相乘")

    def __rmul__(self, other):
        #标量*向量
        return self.__mul__(other)


#进行测试
print("面向对象的Vector: ")
v1 = Vector(2, (2, 3))
v2 = Vector(2, (4, 5))
print("v1:", v1)
print("v2:", v2)
print("v1 == v2:", v1 == v2)
print("v1 < v2:", v1 < v2)
print("v1.dot(v2):", v1.dot(v2))
print("2 * v1:", 2 * v1)
print("v1 * v2:", v1 * v2)
```
